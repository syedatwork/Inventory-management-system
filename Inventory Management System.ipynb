```python
# 1. INITIALIZE DATA STRUCTURE (List of Dictionaries)
# saving inventory details out of the other code, so that we can add and update the inventory through code
inventory = [
    {"name": "Laptop", "price": 55000, "stock": 12},
    {"name": "Smartphone", "price": 30000, "stock": 25},
    {"name": "Headphones", "price": 2500, "stock": 40},
    {"name": "Keyboard", "price": 1200, "stock": 35},
    {"name": "Mouse", "price": 800, "stock": 50},
    {"name": "Monitor", "price": 15000, "stock": 18},
    {"name": "Printer", "price": 10000, "stock": 10},
    {"name": "Tablet", "price": 22000, "stock": 20},
    {"name": "External Hard Drive", "price": 6000, "stock": 15},
    {"name": "USB Flash Drive", "price": 700, "stock": 60},
    {"name": "Smartwatch", "price": 12000, "stock": 22},
    {"name": "Speakers", "price": 3500, "stock": 30},
    {"name": "Webcam", "price": 2000, "stock": 28},
    {"name": "Router", "price": 4000, "stock": 16},
    {"name": "Power Bank", "price": 1500, "stock": 45}
]
```

```python
def display_menu():
    """Prints the main menu options."""
    print("\n--- Inventory Management CLI ---")
    print("1. View All Inventory")
    print("2. Add New Product")
    print("3. Update Stock (Add/Remove)")
    print("4. Calculate Total Inventory Value")
    print("5. Exit")
    print("--------------------------------")

def view_inventory():
    """Loops through the inventory and prints details using string formatting."""
    print("\n--- Current Inventory ---")
    if not inventory:
        print("Inventory is currently empty.")
        return

    # Using f-strings and alignment for a neat table format
    # Concepts Used: String Formatting, Iteration
    print(f"| {'Name':<25} | {'Price (USD)':>15} | {'Stock':>8} |")
    print("-" * 57)
    for product in inventory:
        # Data access (dictionary keys)
        name = product['name']
        price = product['price']
        stock = product['stock']
        
        # String slicing for names too long for the column 
        display_name = name[:24] + (name[24:] and '...') 
        
        print(f"| {display_name:<25} | {price:>15.2f} | {stock:>8} |")

def add_product():
    """Adds a new product to the inventory."""
    print("\n--- Add New Product ---")
    
    # Concepts Used: String Input, String Methods, Type Conversion, List Method (append)
    name = input("Enter product name: ").strip().title()
    
    # Check if product already exists (Logical and Comparison Operators)
    if any(p['name'].lower() == name.lower() for p in inventory):
        print(f"Error: Product '{name}' already exists in inventory.")
        return

    try:
        # Type Conversion and Validation
        price = float(input("Enter unit price (e.g., 59.99): "))
        stock = int(input("Enter initial stock quantity (integer): "))

        if price <= 0 or stock < 0:
            # Conditional Statement
            print("Error: Price must be positive and stock cannot be negative.")
            return

    except ValueError:
        # Error Handling for invalid input types
        print("Error: Invalid input for price or stock. Please ensure they are numbers.")
        return

    # Add the new dictionary to the list
    new_product = {'name': name, 'price': price, 'stock': stock}
    inventory.append(new_product)
    print(f"Success: '{name}' added to inventory.")

def update_stock():
    """Finds a product and updates its stock level."""
    print("\n--- Update Stock ---")
    
    # Concepts Used: String Methods (.lower()), Comparison Operators, Arithmetic Operators
    search_name = input("Enter the name of the product to update: ").strip().lower()
    product_found = False

    for product in inventory:
        if product['name'].lower() == search_name:
            product_found = True
            
            # Use Type Conversion and try/except for safe input
            try:
                quantity_change = int(input("Enter change in stock (use positive for add, negative for remove): "))
            except ValueError:
                print("Error: Invalid input. Please enter an integer.")
                return

            # Arithmetic and Conditional Operators
            new_stock = product['stock'] + quantity_change
            
            if new_stock < 0:
                # Conditional Statement
                print(f"Error: Cannot remove {abs(quantity_change)}. Stock would become negative ({new_stock}).")
                print(f"Current stock for {product['name']}: {product['stock']}")
                return

            # Update the stock value in the dictionary
            product['stock'] = new_stock
            print(f"Success: Stock for {product['name']} updated.")
            print(f"New stock level: {product['stock']}")
            break
            
    if not product_found:
        print(f"Error: Product '{search_name.title()}' not found in inventory.")

def calculate_value():
    """Calculates the total monetary value of all items in stock."""
    # Concepts Used: Arithmetic Operators, Iteration, Variable Accumulation
    total_value = 0.0
    
    if not inventory:
        print("\nInventory is empty, total value is $0.00.")
        return

    for product in inventory:
        # Arithmetic Operator for calculation: price * stock
        item_value = product['price'] * product['stock']
        
        # Accumulation
        total_value += item_value

    # String Formatting for currency display
    print("\n--- Inventory Value Summary ---")
    print(f"Total value of all items in stock: ${total_value:,.2f}") # .2f gives the ',' every after 2 digits

# --- MAIN PROGRAM LOOP ---
# Concept Used: Conditional Statements (if/elif/else) and Loop (while True)

def main():
    while True:
        display_menu()
        
        # Get user input (String Input)
        choice = input("Enter your choice (1-5): ").strip()

        if choice == '1':
            view_inventory()
        elif choice == '2':
            add_product()
        elif choice == '3':
            update_stock()
        elif choice == '4':
            calculate_value()
        elif choice == '5':
            # Exit the loop (Special Operator: break)
            print("\nExiting Inventory Management System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 5.")

# Run the main function
if __name__ == "__main__":
    main()
```
